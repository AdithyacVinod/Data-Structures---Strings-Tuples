```python
# String Concatenation

string1 = "Hello "
name = "Adithya"

string2 = string1 + name
print(string2)

string3 = ", welcome to Python programming"
string2 = string2 + string3

print(string2)
```



* operator "+" is used for string concatenation, which joins two or   more strings together.
* string1 + name combines "Hello " and "Adithya".
  
* Then string2 + string3 adds the welcome message to the existing string.








```python
# String Slicing and Indexing

text = "Hello Adithya, welcome to Python programming"

# a. First character
print("First character:", text[0])

# b. Last character
print("Last character:", text[-1])

# c. First 5 characters
print("First 5 characters:", text[:5])

# d. Last 11 characters
print("Last 11 characters:", text[-11:])

# e. Reverse the string
print("Reverse:", text[::-1])

# f. Print the word Python
print("Word Python:", text[28:34])
```



* Indexing is used to access a specific character from a string.
* text[0] gives the first character.
* text[-1] gives the last character.
* Slicing is used to extract a part of a string.
* text[:5] extracts the first 5 characters.
* text[-11:] extracts the last 11 characters.
* text[::-1] reverses the complete string.
* text[28:34] extracts the word Python using its index positions.










```python
# String Methods

strM = "Python beginner tutorial"

# a. Convert to uppercase
print("Uppercase:", strM.upper())

# b. Convert to lowercase
print("Lowercase:", strM.lower())

# c. Capitalize the sentence
print("Capitalized:", strM.capitalize())

# d. Count the occurrences of 't'
print("Count of 't':", strM.count('t'))

# e. Replace Python with Machine Learning
print("After replacement:", strM.replace("Python", "Machine Learning"))
```



* .upper() converts all characters into uppercase.
* .upper() converts all characters into uppercase.
* .capitalize() makes the first character uppercase and the remaining characters lowercase.
* .count() counts how many times a particular character or word  occurs.
* .replace() replaces a specified word or character with another value.




```python
# Tuples

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples
t_combine = tuple1 + tuple2
print("Combined tuple:", t_combine)

# b. Repeat t_combine 3 times
t_repeat = t_combine * 3
print("Repeated tuple:", t_repeat)

# c. Access the 3rd element
print("3rd element:", t_combine[2])

# d. Access the first three elements
print("First three elements:", t_combine[:3])

# e. Access the last three elements
print("Last three elements:", t_combine[-3:])
```



* Tuple is an ordered collection of elements enclosed in ().
* Operator "+" is used to concatenate two tuples.
* Operator "*" repeats the elements of a tuple.
* Indexing is used to access a particular element. t_combine[2] gives the 3rd element because Python indexing starts from 0.
* Slicing is used to access a range of elements.
* t_combine[:3] gives the first three elements.
* t_combine[-3:] gives the last three elements.
